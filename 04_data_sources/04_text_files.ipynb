```python
from pyspark.sql.session import SparkSession


spark = (

    SparkSession.builder.getOrCreate()
)

spark
```

```python
text_df  = (
    spark
    .read
    .text("../data/flight-data/csv")
)

text_df.show(truncate=False)
```

```python
from pyspark.sql.functions import split

splitted_df = (

    text_df
    .withColumn("city",split("value",",")[0])
)

splitted_df.show()
```

```python
splitted_df.write.format("text").partitionBy("city").save("../data/tmp/text-file")
```

```python
spark.stop()
```
